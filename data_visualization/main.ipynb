```python
import matplotlib.pyplot as plt

dataX = [1, 2, 3, 4, 5]
dataY = [2, 5, 7, 4, 13]

plt.plot(dataX, dataY, marker='o')
plt.title('Contoh plot sederhana')
plt.xlabel('Nilai x')
plt.ylabel('Nilai y')
plt.grid(True)
plt.show
```

```python
import numpy as np
x = np.random.rand(50)
y = np.random.rand(50)

colors = np.random.rand(50)
sizes = 500 * np.random.rand(50)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='brg')
plt.colorbar()
plt.title('scatter plot')
plt.xlabel('Nilai x')
plt.ylabel('Nilai y')
plt.show()
```

```python
kategori = ['A', 'B', 'C', 'D', 'E']
nilai = [2, 3, 7, 4, 8]

plt.bar(kategori, nilai, color='blue')
plt.title('Bar Cari Sederhana')
plt.xlabel('Kategori')
plt.ylabel('Nilai')
plt.show
```

```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['Bagian A', 'Bagian B', 'Bagian C', 'Bagian D']
sizes = [215, 100, 200,150]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Pie Chart Distribusi Bagian')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data_saham = pd.read_csv('ANTM.csv')
plt.figure(figsize=(10, 6))
sns.distplot(data_saham['close'])
plt.title('Distribusi harga penutupan saham')
plt.xlabel('Harga Penutupan')
plt.ylabel('Frekuensi')
plt.show
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axs = plt.subplots(2, 2, figsize=(10, 10))

axs[0, 0].plot([1, 2, 3], [3, 2, 1], 'r')
axs[0, 0].set_title('Subplot 1-1')

axs[0, 1].bar([1, 2, 3], [3, 2, 1], color='r')
axs[0, 1].set_title('Subplot 1-2')

axs[1, 0].hist(np.random.randn(100), bins=20)
axs[1, 0].set_title('Subplot 2-1')

axs[1, 1].scatter(np.random.randn(50), np.random.rand(50))
axs[1, 1].set_title('Subplot 2-2')

plt.tight_layout()
plt.show()
```
